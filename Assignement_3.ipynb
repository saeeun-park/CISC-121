Using this idea, I implemented a function that repeatedly checks the middle version, narrows the search to the left if it is bad, or to the right if it is not, until it finds the first bad version.


```python
def firstBadVersion(n):
    left = 1
    right = n
    while left <= right:
        mid = (left + right) // 2
        if isBadVersion(mid):
            right = mid - 1
        else:
            left = mid + 1
    return left
```

```python
def isAnagram(s, t):
    if len(s) != len(t):
        return False

    counter = {}
    for char in s:
        counter[char] = counter.get(char, 0) + 1

    for char in t:
        if char not in counter:
            return False
        counter[char] -= 1

    return False
```

```python
def isAnagram(s, t):
    if len(s) != len(t):
        return False

    counter = {}
    for char in s:
        counter[char] = counter.get(char, 0) + 1

    for char in t:
        if char not in counter:
            return False
        counter[char] -= 1

    return False
```
